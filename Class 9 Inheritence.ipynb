#### Q1. Vehicle→Car: override display_info() and use super().####  Requirements: Use inheritance, constructors in both classes, super().__init__(), override the method with the same name, and create objects to demonstrate both parent and child behavior.

 ####  Requirements: Use inheritance, constructors in both classes, super().__init__(), override the method with the same name, and create objects to demonstrate both parent and child behavior.

```python
class HospitalStaff:
    def __init__(self, name):
        self.name = name
        print("HospitalStaff constructor called")

    def work(self):
        print(self.name, "works in the hospital.")


class Doctor(HospitalStaff):
    def __init__(self, name, specialty):
        super().__init__(name)
        self.specialty = specialty
        print("Doctor constructor called")

    def work(self):
        print(self.name, "works as a", self.specialty)


s1 = HospitalStaff("Ali")
s1.work()

print()

s2 = Doctor("Dr. Ahmed", "Cardiologist")
s2.work()
```

#### Q39. Delivery→ExpressDelivery: override deliver().
 ####  Requirements: Use inheritance, constructors in both classes, super().__init__(), override the method with the same name, and create objects to demonstrate both parent and child behavior.

```python
class Delivery:
    def __init__(self, company):
        self.company = company
        print("Delivery constructor called")

    def deliver(self):
        print(self.company, "delivers parcels normally.")


class ExpressDelivery(Delivery):
    def __init__(self, company, time):
        super().__init__(company)
        self.time = time
        print("ExpressDelivery constructor called")

    def deliver(self):
        print(self.company, "delivers parcels within", self.time)


d1 = Delivery("TCS")
d1.deliver()

print()

d2 = ExpressDelivery("Leopards", "24 Hours")
d2.deliver()
```

#### Q40. MusicPlayer→SpotifyPlayer: override play_music().
  #### Requirements: Use inheritance, constructors in both classes, super().__init__(), override the method with the same name, and create objects to demonstrate both parent and child behavior.

```python
class MusicPlayer:
    def __init__(self, name):
        self.name = name
        print("Music player constructor is called: ")

    def playMusic(self):
        print(self.name, "plays the music: ")

class SpotifyPlayer(MusicPlayer):
    def __init__(self, name, music_type):
        super().__init__(name)
        self.music_type = music_type
        print("Spotifyplayer constructor is called: ")

    def playMusic(self):
        super().playMusic()
        print(self.name,"Enjoys", self.music_type , "in party:")

v1 = MusicPlayer("John")
v1.playMusic()

print()

v2 = SpotifyPlayer("Williams", "Pop music")
v2.playMusic()
```

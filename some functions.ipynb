## **Max() Function**
returns the largest value

```python
marks = (21,34,43,42.66,56,98)
print("highest value:", max(marks))
```

## **Min() Function**
returns the smallest value

```python
marks = (21,34,43,42.66,56,98)
print("smallest value :", min(marks))
```

## **Sum() Function**
return the sum value

```python
marks = (21,34,43,42,66,56,98)
print("sum value :", sum(marks))
```

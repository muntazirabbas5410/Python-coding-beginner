## **Tuple**
immutable (unchangeable)

```python
fruit = ("apple", "banana", "cherry")
print(fruit)
print(type(fruit))
```

### **Tuple unpacking**
we can unpack tuple elements into variables

```python
coordinates = (10,20)
x,y = coordinates
print(x)
print(y)
```

### **Nesting Tuples**
Tuples can contain other tuples

```python
nested = ((1,2), (3,4), (5,6))
print(nested[0])
print(nested[0][1])

print(nested[1])
print(nested[1][0])

print(nested[2])
print(nested[2][1])
```

### **Tuple Methods**
###count()-Returns the number of occuurences of a value .
###index()-Returns the index of the first occurence of a value.

```python
fruit = ("apple", "banana", "cherry", "mango", "kiwi")
print(fruit.count("apple"))
print(fruit.index("cherry"))
```

### **Checking Membership**
use in and not in to check if an element exist in tuple:

```python
fruit = ("apple", "banana", "cherry", "mango", "kiwi")
print("apple" in fruit)
print("mango" in fruit)
```

### **Sorting Tuples**
we can't sort a tuple directly, but we can create a sorted list from it:


```python
numbers = (1,2,3,4,5)
sorted_numbers = sorted(numbers)
print(sorted_numbers)
```

```python
num = (1,2,3,4,5)
sorted(num)
print(num)
```

### **Convert tuple to list and vice versa**
we can convert a tuple to a list and back:

```python
fruit = ("apple", "banana", "cherry", "mango", "kiwi")
fruit_list = list(fruit)
print(fruit_list)

print_tuple = tuple(fruit_list)
print(print_tuple)
```

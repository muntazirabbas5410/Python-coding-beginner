**Q1. Write a program that asks the user to enter a password repeatedly until they enter 'python123'**.

```python
password = ""

while password != "python123":
    password = input("Enter the password: ")

print("Login")
```

**Q2. Ask the user to enter positive numbers continuously. Stop when the user enters -1 and print all numbers entered.**

```python
user = int(input("Enter the positive number"))
while user != -1:
    print(user)
    user = int(input("Enter the positive number"))
```

**Q3. Using a while loop, ask the user to enter five favorite movies, store them in a list, and print the list.**

```python
movies = []
count = 1

while count <= 5:
    movie = input(f"Enter favorite movie {count}: ")
    movies.append(movie)
    count += 1

print("Your favorite movies are:")
print(movies)
```

**Q4. Ask the user to enter student marks until -1 is entered. Print Pass or Fail for each mark.**

```python
student = int(input("enter the marks"))
while student != -1:
    if student >= 40:
        print("Pass")
    else:
        print("Fail")
    student = int(input("enter the marks"))
```

**Q5. Create a tuple of weekdays and print each day using a while loop and len().**

```python
weeks = ("Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday")
count = 0

while count < len(weeks):
    print(weeks[count])
    count += 1
```

**Q6. Create a dictionary containing Name, Age, and City. Using a list of keys and a while loop, print all values.**

```python
person = {
    "Name": "Ali",
    "Age": 22,
    "City": "Karachi"
}

keys = ["Name", "Age", "City"]

i = 0

while i < len(keys):
    print(person[keys[i]])
    i += 1
```

**Q7. Ask the user to enter five numbers using a while loop, store them in a list, and print only the even numbers.**

```python
user = int(input("enter the numbers"))
count = 1

while count <=5:
  if user % 2 == 0:
    print(user)
  user = int(input("enter the numbers"))
  count += 1
```

**Q8. Create a list of fruits and print each fruit in uppercase using a while loop.**

```python
fruits = ["kiwi", "apple", "mango", "orange"]
i = 0
while i < len(fruits):
    print(fruits[i].upper())
    i += 1
```

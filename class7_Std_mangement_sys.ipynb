### **Student Record Management System**

```python
student = [
    {"Name" : "Muntazir Abbas", "Age" : 23, "marks" : 1000},
    {"Name" : "Abbas", "Age" : 22, "marks" :950},
    {"Name" : "khurasani", "Age" : 33, "marks" : 870}
]
print("Initial students:", student)

subject = ("physics", "maths", "physics", "computer")


my_list = []

while True:
  print("========student management system========")
  print("1. add student")
  print("2. view student")
  print("3. view subjects")
  print("4. exit")

  choice = input("enter your choice:")
  if choice == "1":
    name = input("enter student name:")
    age = int(input("enter student age:"))
    marks = int(input("enter student marks:"))

    new_student = {
        "Name" : name,
        "Age" : age,
        "marks" : marks
    }

    student.append(new_student)
    print("student added successfully")
  elif choice == "2":
    if len(student) == 0:
      print("no student found")
    else:
      print("\n--- Student List ---")
      for s in student:
        print(f"Name: {s['Name']}, Age: {s['Age']}, Marks: {s['marks']}")
  elif choice == "3":
    print("\n--- Subjects ---")
    for sub_name in subject:
        print(sub_name)
  elif choice == "4":
    print("Exiting student management system. Goodbye!")
    break
  else:
    print("Invalid choice. Please enter a number between 1 and 4.")
```

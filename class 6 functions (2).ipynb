Q17. Create a function that receives a list of fruits and prints each fruit.

```python
def list_fruits(fruits):
    for fruit in fruits:
        print(fruit)

fruits_list = ["kiwi", "apple", "orange", "banana", "mango"]

list_fruits(fruits_list)
```

Q18. Create a function that receives a tuple of countries and prints each country.

```python
def country_list(countries):
    for country in countries:
        print(country)

list_country = ("Pakistan", "India", "Nepal", "Germany", "Australia")

country_list(list_country)
```

Q19. Create a function that receives a student dictionary and prints all values.

```python
def student_values(values):
    for value in values.values():
        print(value)

values = {
    "Name": "Muntazir Abbas",
    "CGPA": 3.9,
    "Department": "Data Science"
}

student_values(values)
```

Q20. Create a function that prints numbers from 1 to n using a for loop.

```python
def number(n):
    for i in range(1, n+1):
        print(i)

number(20)
```

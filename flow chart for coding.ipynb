## In Python, data types are commonly divided into Primitive (Basic) and Container (Collection) data types.

### ***1. Primitive dataypes*** :These store a single value.
###(int, float, boolean, complex, string)
###They are the basic building blocks of every Python program and are used because they are simple, fast, and efficient. Also provide the basic data types needed for calculations, text handling, and decision-making in a program.


### ***2.  Container (Collection) Data Types*** : These store multiple values.      
###(list, tuple, set, dict)
###They help organize, manage, and process collections of data efficiently instead of creating many separate variables.and organize multiple values in a single variable, making data easier to manage and process.

### **Slicing**

```python
A = ["apple" , "banana" , "kiwi" , "orange" , "mango"]
A[1:3] = ["pineapple" , "strawberry"]
print(A)
```

```python
NAME = ['M','U','N','T','A','Z','I','R']
reverse_name = NAME[::-1]

print(reverse_name)
```

### pop()

### **Remove()**

```python
students = []

while True:

    print("\n======== Students Attendance System ========")
    print("1. Add Student")
    print("2. View Students")
    print("3. Mark Attendance")
    print("4. View Attendance")
    print("5. Search Student")
    print("6. Update Student")
    print("7. Delete Student")
    print("8. Exit")

    choice = input("Enter your choice: ")

    #Add Student
    if choice == "1":

        ID = int(input("Enter ID: "))
        name = input("Enter Name: ")
        roll_no = int(input("Enter Roll No: "))
        class_name = input("Enter Class: ")
        section = input("Enter Section: ")
        semester = int(input("Enter Semester: "))

        student = {
            "ID": ID,
            "Name": name,
            "Roll No": roll_no,
            "Class": class_name,
            "Section": section,
            "Semester": semester,
            "Attendance": "Absent"
        }

        students.append(student)

        print("Student Added Successfully!")

    #View Students 
    elif choice == "2":

        if len(students) == 0:
            print("No Students Found!")

        else:
            print("\nStudent List")

            for student in students:
                print(student)

    #Mark Attendance
    elif choice == "3":

        ID = int(input("Enter Student ID: "))

        found = False

        for student in students:

            if student["ID"] == ID:

                status = input("Present / Absent : ")

                student["Attendance"] = status

                print("Attendance Updated Successfully!")

                found = True
                break

        if found == False:
            print("Student Not Found!")

    #View Attendance
    elif choice == "4":

        if len(students) == 0:

            print("No Students Found!")

        else:

            for student in students:

                print(student["Name"], ":", student["Attendance"])

    #Search Student
    elif choice == "5":

        ID = int(input("Enter Student ID: "))

        found = False

        for student in students:

            if student["ID"] == ID:

                print(student)

                found = True
                break

        if found == False:

            print("Student Not Found!")

    #Update Student
    elif choice == "6":

        ID = int(input("Enter Student ID: "))

        found = False

        for student in students:

            if student["ID"] == ID:

                student["Name"] = input("New Name: ")
                student["Roll No"] = int(input("New Roll No: "))
                student["Class"] = input("New Class: ")
                student["Section"] = input("New Section: ")
                student["Semester"] = int(input("New Semester: "))

                print("Student Updated Successfully!")

                found = True
                break

        if found == False:

            print("Student Not Found!")

    #Delete Student
    elif choice == "7":

        ID = int(input("Enter Student ID: "))

        found = False

        for student in students:

            if student["ID"] == ID:

                students.remove(student)

                print("Student Deleted Successfully!")

                found = True
                break

        if found == False:

            print("Student Not Found!")

    #Exit
    elif choice == "8":

        print("Program Closed Successfully!")

        break

    #Invalid Choice
    else:

        print("Invalid Choice! Please Try Again.")
```

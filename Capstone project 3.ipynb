### Hospital management system:

```python
from abc import ABC, abstractmethod


# Abstract Parent Class
class Employee(ABC):   #Employee abstract parent class hai.

    def __init__(self, name, employee_id, salary, experience):
        self.name = name
        self.__employee_id = employee_id
        self.__salary = salary
        self.__experience = experience

    # Getter methods
    def get_employee_id(self):   #Inhein (emp,sal,exp)directly access nahi kiya ja sakta. Isliye getters use kiye:
        return self.__employee_id

    def get_salary(self):
        return self.__salary

    def get_experience(self):
        return self.__experience

    # Abstract methods
    @abstractmethod
    def perform_duty(self):
        pass

    @abstractmethod
    def calculate_salary(self):
        pass


# Doctor Class
class Doctor(Employee):

    def __init__(self, name, employee_id, salary, experience, specialization):
        super().__init__(name, employee_id, salary, experience)
        self.specialization = specialization

    def perform_duty(self):
        print(f"Doctor {self.name} is examining patients.")

    def calculate_salary(self):
        salary = self.get_salary() + (self.get_experience() * 5000)
        print(f"Doctor Salary: {salary}")


# Nurse Class
class Nurse(Employee):

    def __init__(self, name, employee_id, salary, experience):
        super().__init__(name, employee_id, salary, experience)

    def perform_duty(self):
        print(f"Nurse {self.name} is taking care of patients.")

    def calculate_salary(self):
        salary = self.get_salary() + (self.get_experience() * 3000)
        print(f"Nurse Salary: {salary}")


# Receptionist Class
class Receptionist(Employee):

    def __init__(self, name, employee_id, salary, experience):
        super().__init__(name, employee_id, salary, experience)

    def perform_duty(self):
        print(f"Receptionist {self.name} is managing the reception.")

    def calculate_salary(self):
        salary = self.get_salary() + (self.get_experience() * 2000)
        print(f"Receptionist Salary: {salary}")


# Pharmacist Class
class Pharmacist(Employee):

    def __init__(self, name, employee_id, salary, experience):
        super().__init__(name, employee_id, salary, experience)

    def perform_duty(self):
        print(f"Pharmacist {self.name} is managing medicines.")

    def calculate_salary(self):
        salary = self.get_salary() + (self.get_experience() * 2500)
        print(f"Pharmacist Salary: {salary}")


# Creating Objects
doctor = Doctor("Ali","D001", 100000,5,"Cardiologist")

nurse = Nurse("Zehra","N001",60000,3)

receptionist = Receptionist("Ahmed","R001",50000, 2)

pharmacist = Pharmacist(  "Fatima",  "P001",  70000, 4)


# Polymorphism
employees = [doctor, nurse, receptionist, pharmacist]

for employee in employees:
    print("-------------------------")
    print("Employee:", employee.name)
    print("Employee ID:", employee.get_employee_id())
    print("Experience:", employee.get_experience(), "years")

    employee.perform_duty()
    employee.calculate_salary()
```
